```python
text = input("Enter words separated by space: ")

words = text.split()

for word in words:

    tag = "NOUN"

    if word.endswith("ing"):
        tag = "VERB"

    elif word.endswith("ed"):
        tag = "VERB"

    elif word.endswith("ly"):
        tag = "ADVERB"

    print(word, "->", tag)
```
