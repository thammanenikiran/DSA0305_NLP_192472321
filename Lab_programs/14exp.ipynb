```python
sentence = input("Enter sentence: ")

words = sentence.lower().split()

subject = words[0]
verb = words[1]

if subject in ["he", "she", "it"] and verb.endswith("s"):
    print("Agreement is Correct")

elif subject in ["they", "we", "you"] and not verb.endswith("s"):
    print("Agreement is Correct")

else:
    print("Agreement is Incorrect")
```
