```python
from collections import Counter
import math

text = """
Machine learning is a branch of artificial intelligence.
Artificial intelligence allows computers to learn from data.
Machine learning algorithms use data to make predictions.
Predictions can help organizations make better decisions.
"""

sentences = text.strip().split("\n")

def similarity(sentence1, sentence2):

    words1 = set(sentence1.lower().split())
    words2 = set(sentence2.lower().split())

    common = len(words1.intersection(words2))
    total = len(words1.union(words2))

    if total == 0:
        return 0

    return common / total


total_score = 0
count = 0

print("Sentence Similarity:")

for i in range(len(sentences) - 1):

    score = similarity(sentences[i], sentences[i + 1])

    print("Sentence", i + 1, "and Sentence", i + 2,
          "=", round(score, 2))

    total_score = total_score + score
    count = count + 1

coherence = total_score / count

print("\nCoherence Score:", round(coherence, 2))

if coherence >= 0.15:
    print("Text is Coherent")
else:
    print("Text is Less Coherent")
```
