```python
noun = input("Enter a singular noun: ")

if noun.endswith(("s", "x", "z", "ch", "sh")):
    plural = noun + "es"

elif noun.endswith("y") and noun[-2].lower() not in "aeiou":
    plural = noun[:-1] + "ies"

else:
    plural = noun + "s"

print("Singular:", noun)
print("Plural:", plural)
```
