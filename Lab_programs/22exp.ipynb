```python
import nltk

nltk.download('punkt')

text = "John went to the market. He bought a book. It was interesting."

sentences = text.split(".")

previous_noun = ""

print("Text:", text)
print("\nReference Resolution:")

for sentence in sentences:

    words = sentence.strip().split()

    if len(words) == 0:
        continue

    for word in words:

        if word in ["John", "Mary", "book", "car", "student"]:
            previous_noun = word

        if word in ["He", "he", "She", "she", "It", "it"]:
            print(word, "->", previous_noun)
```
