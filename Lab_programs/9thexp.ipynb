```python
import re

text = input("Enter words separated by space: ")

words = text.split()

for word in words:

    if re.match(r"^[0-9]+$", word):
        tag = "NUMBER"

    elif word.endswith("ing"):
        tag = "VERB"

    elif word.endswith("ly"):
        tag = "ADVERB"

    elif word.endswith("ous"):
        tag = "ADJECTIVE"

    else:
        tag = "NOUN"

    print(word, "->", tag)
```
