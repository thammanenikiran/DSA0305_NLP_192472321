```python
dialog = [
    "Hello",
    "What is your name?",
    "My name is Kiran.",
    "Can you help me?",
    "Sure, I can help you.",
    "Thank you",
    "Goodbye"
]

print("Dialog Act Recognition:\n")

for sentence in dialog:

    text = sentence.lower()

    if "hello" in text or "hi" in text:
        act = "Greeting"

    elif "?" in sentence:
        act = "Question"

    elif "can you" in text or "please" in text:
        act = "Request"

    elif "thank" in text:
        act = "Thanking"

    elif "goodbye" in text or "bye" in text:
        act = "Goodbye"

    elif "sure" in text:
        act = "Answer"

    else:
        act = "Statement"

    print(sentence, "->", act)
```
