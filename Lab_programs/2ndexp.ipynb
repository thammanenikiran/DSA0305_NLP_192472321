```python
text = input("Enter a string: ")

state = "q0"

for ch in text:

    if state == "q0":
        if ch == "a":
            state = "q1"
        else:
            state = "q0"

    elif state == "q1":
        if ch == "b":
            state = "q2"
        elif ch == "a":
            state = "q1"
        else:
            state = "q0"

    elif state == "q2":
        if ch == "a":
            state = "q1"
        else:
            state = "q0"

if state == "q2":
    print("String Accepted")
else:
    print("String Rejected")
```
