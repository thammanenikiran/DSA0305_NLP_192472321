```python
import nltk

nltk.download("punkt")

sentence = input("Enter sentence: ")

grammar = nltk.CFG.fromstring("""
S -> NP VP
NP -> Det N
VP -> V NP
Det -> 'the' | 'a'
N -> 'boy' | 'girl' | 'dog' | 'cat'
V -> 'sees' | 'likes'
""")

words = sentence.lower().split()

parser = nltk.ChartParser(grammar)

trees = list(parser.parse(words))

if trees:
    print("Parse Tree:")
    trees[0].pretty_print()
else:
    print("No parse tree found")
```
