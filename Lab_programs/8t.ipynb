```python
text = input("Enter a sentence: ")

words = text.lower().split()

tags = {
    "the": "DT",
    "a": "DT",
    "boy": "NN",
    "girl": "NN",
    "dog": "NN",
    "runs": "VB",
    "plays": "VB",
    "beautiful": "JJ"
}

print("Word\tPOS")
print("----------------")

for word in words:

    if word in tags:
        print(word, "\t", tags[word])
    else:
        print(word, "\t", "NN")
```
