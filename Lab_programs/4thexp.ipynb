```python
noun = input("Enter a singular noun: ")

length = 0

for ch in noun:
    length = length + 1

last = noun[length - 1]

if last == "s" or last == "x" or last == "z":
    plural = noun + "es"

elif length >= 2:

    second_last = noun[length - 2]

    if (last == "h" and
        (second_last == "c" or second_last == "s")):
        plural = noun + "es"

    elif last == "y":

        if (second_last != "a" and
            second_last != "e" and
            second_last != "i" and
            second_last != "o" and
            second_last != "u"):

            plural = noun[0:length-1] + "ies"

        else:
            plural = noun + "s"

    else:
        plural = noun + "s"

else:
    plural = noun + "s"

print("Singular:", noun)
print("Plural:", plural)
```
